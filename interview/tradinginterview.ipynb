```python
# Knockout Stage
# Probability and Statistics Theory, General, Medium

# 32 teams are playing in a competition. They are currently at the knock out stage, 
# where each game 2 teams compete with each other to continue to the next round. 
# The teams are numbered from 1 to 32, according to their superiority. In other words, 
# Team #1 always beats Team #2, Team #2 always beats Team #3, etc. 
# What's the probability that Team #1 and Team #3 will play in the final?

import numpy as np

def get_record(teams):
    # shuffle teams
    np.random.shuffle(teams)

    # get first and second half
    left = teams[:17]
    right = teams[17:]

    if (
            (1 in left) and (2 in left) and (3 in right)
    ) or (
            (1 in right) and (2 in right) and (3 in left)
    ):
        return True
    else:
        return False
    
def estimate_probability(seed = 12345, num_rounds = 1000):
    np.random.seed(seed)

    teams = np.array([i for i in range(1, 33)])

    total_records = 0

    for round in range(num_rounds):
        # get record
        total_records += 1 * get_record(teams)

    return total_records / num_rounds

estimate_probability()
```

```python
# Given a stream of stock data, create a data structure that can efficiently handle the following functions:

# 1. execute_trade(ticker, volume) - store the trade that has occurred
# 2. most_traded(k) - return the k most traded stocks by volume. The return format should be a list of strings with each string being formatted as: <TICKER> <VOLUME>

from typing import List

class StockTracker:
    def __init__(self) -> None:
        self.d = dict()
        self.l = []

    def execute_trade(self, ticker: str, volume: int) -> None:
        if ticker not in self.d:
            self.d[ticker] = len(self.l)
            self.l.append([0, ticker])
        
        # get index in l
        idx = self.d[ticker]

        # update ticker's volume in l
        self.l[idx][0] += volume

        # bubble up / down based on volume
        while (idx >= 1) and (self.l[idx][0] > self.l[idx - 1][0]):
            self.l[idx], self.l[idx - 1] = self.l[idx - 1], self.l[idx]
            idx -= 1

        while (idx < len(self.l) - 1) and (self.l[idx][0] < self.l[idx + 1][0]):
            self.l[idx], self.l[idx + 1] = self.l[idx + 1], self.l[idx]
            idx += 1

        # update index in d
        self.d[ticker] = idx


    def most_traded(self, k: int) -> List[str]:
        ret = ""

        for i in range(min(k, len(self.l))):
            ret += f"{self.l[i][1]} {self.l[i][0]}\n"

        print(ret)
    

st = StockTracker()

st.execute_trade('TSLA', 1000)
st.execute_trade('NFLX', 700)
st.execute_trade('TSLA', 200)
st.execute_trade('META', 1400)
st.most_traded(2)
```

```python
from typing import List
import heapq

class StockTracker:
    def __init__(self) -> None:
        self.d = dict()

    def execute_trade(self, ticker: str, volume: int) -> None:
        self.d[ticker] = self.d.get(ticker, 0) + volume


    def most_traded(self, k: int) -> List[str]:
        h = []

        for ticker, volume in self.d.items():
            heapq.heappush(h, [volume, ticker])

            if len(h) > k:
                heapq.heappop(h)

        ret = "\n".join([f"{item[1]} {item[0]}" for item in h[::-1]])

        print(ret)
    

st = StockTracker()

st.execute_trade('TSLA', 1000)
st.execute_trade('NFLX', 700)
st.execute_trade('TSLA', 200)
st.execute_trade('META', 1400)
st.most_traded(2)
```
