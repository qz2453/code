```python
import numpy as np

def getProfit(s1, s2):
    n = len(s1)

    dp = np.zeros((n, 3, 3, 2, 2))
    dp[0, 0, 0, 0, 0] = 0
    dp[0, 1, 0, 1, 0] = -s1[0]
    dp[0, 0, 1, 0, 1] = -s2[0]

    for i in range(1, n):
        for j1 in range(3):
            for j2 in range(3):
                if j1 > 0:
                    dp[i, j1, j2, 1, 0] = max(dp[i - 1, j1, j2, 1, 0], dp[i - 1, j1 - 1, j2, 0, 0] - s1[i])

                if j2 > 0:
                    dp[i, j1, j2, 0, 1] = max(dp[i - 1, j1, j2, 0, 1], dp[i - 1, j1, j2 - 1, 0, 0] - s2[i])

                dp[i, j1, j2, 0, 0] = max(dp[i - 1, j1, j2, 0, 0], dp[i - 1, j1, j2, 1, 0] + s1[i], dp[i - 1, j1, j2, 0, 1] + s2[i])

    return dp[n - 1, 2, 2, 0, 0]


s1 = [3, 3, 5, 0, 0, 3, 1, 4]
s2 = [0, 0, 6, 1, 3, 5, 0, 1]

s1 = [0]
s2 = [0]

getProfit(s1, s2)

```
